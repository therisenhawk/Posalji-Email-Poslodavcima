```python
import pandas as pd


poslodavci = pd.read_excel('pso_poslodavci_pozicije.xlsx')

poslodavci.head(5)
```

```python
poslodavci = poslodavci.copy()
poslodavci.shape
```

```python
#na ove mogu
studijski_programi = [
    "Menadžment informacionih sistema",
    "Informacione tehnologije",
    "Primijenjeno računarstvo",
    "Računarstvo i informacione tehnologije",
    "Računarske nauke",
    "Softverski inženjering",
    "Elektronika, telekomunikacije i računari",
    "Elektrotehnika i računarstvo",
    "Elektronske komunikacije"
]


```

```python
#radna mjesta dje mogu
radna_mjesta = [
    "Analitičar podataka",
    "Data analitičar",
    "Data and Analytics Engineer",
    "AI expert assistant",
    "Artificial Intelligence Project and Product Manager",
    "Artificial Intelligence Project Coordinator",
    "Artificial Intelligence Marketing Coordinator",
    "Saradnik za aktuarske poslove",
    "Sektor za ICT, inovacije i unapređenje poslovanja",
    "Implementator poslovnih rješenja",
    "IT Product Manager",
    "Product owner",
    "Saradnik za informacione sisteme",
    "Saradnik informacionih sistema",
    "Asistent u službi za informacione sisteme i tehnologije",
    "Saradnik u službi za informacione tehnologije",
    "Saradnik na unapređenju i razvoju IT sistema",
    "Saradnik u IT sektoru",
    "Sektor za informacione tehnologije i digitalizaciju",
    "Sektor za razvoj servisa i digitalnu transformaciju",
    "Odjeljenje za razvoj proizvoda i e-business",
    "Pripravnik u sektoru za razvoj proizvoda",
    "Pripravnik u sektoru za unapređenje nabavki",
    "Sektor upravljanja nefinansijskim rizicima",
    "Služba za strateško upravljanje rizicima",
    "Stručni saradnik za marketing i digitalne sisteme",
    "Stručni saradnik u prodaji"
]
```

```python
#vratim redove gdje koja su == radnim mjestima i programima gdje mogu
poslodavci_relevantni = poslodavci.loc[
    (poslodavci['Studijski program'].isin(studijski_programi)) &
    (poslodavci['Naziv radnog mjesta'].isin(radna_mjesta))
].reset_index(drop=True)

poslodavci_relevantni
```
